```python
from bs4 import BeautifulSoup
import requests
import smtplib
import time
import datetime
import re
```

```python
URL = "https://www.amazon.co.za/s?k=sofa&crid=MIHZU8W72X1S&sprefix=sofa%2Caps%2C390&ref=nb_sb_noss_1"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36 Edg/144.0.0.0",
    "Accept-Language": "en-US,en;q=0.9"
}

page = requests.get(URL, headers=headers)
soup = BeautifulSoup(page.content, "html.parser")

# Each product on search results is inside a div with data-component-type="s-search-result"
results = soup.find_all("div", {"data-component-type": "s-search-result"})

data = []
for item in results:
    # Product title can be inside a span with class 'a-text-normal'
    title_tag = item.find("h2")
    title = title_tag.get_text(strip=True) if title_tag else "Title not found"

    # Price is usually inside span with class 'a-offscreen'
    price_tag = item.find("span", class_="a-offscreen")
    price = price_tag.get_text(strip=True) if price_tag else "Price not found"

    data.append([title, price])



```

```python
print(data)
```

```python
import csv

header = ['title', 'price']

with open('AmazonWebScraping.csv', 'w',newline='',encoding='UTF8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(data)
```
